# Shared-depth byte language model analysis
This notebook inspects executed training artifacts; `train.py` contains the complete training pipeline.

```python
import json
from pathlib import Path
import pandas as pd
p = Path('.')
m = json.loads((p / 'metrics.json').read_text())
print('Selected using validation:', m['selected_model'])
```

```python
rows = [{'model': k, **v['test']} for k, v in m['models'].items()]
print(pd.DataFrame(rows).to_string(index=False))
```

```python
e = pd.read_csv(p / 'error_analysis.csv')
for name, v in m['models'].items():
    assert abs(e[e.model == name].nll.mean() - v['test']['nll']) < 1e-7
print('All test NLL values match saved window records.')
```

```python
s = json.loads((p / 'generated_samples.json').read_text())
for sample in s:
    print(sample['model'], repr(sample['prompt']), repr(sample['generated']))
```
